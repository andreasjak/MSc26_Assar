```python
import duckdb
import pandas as pd
import numpy as np
import torch
import torch.nn as nn
from torch.utils.data import DataLoader, TensorDataset
import re
from sklearn.preprocessing import StandardScaler
import sys
sys.path.append('../functions')
from functions.process_data import pre_process_data

print("Beginning processing")
con = duckdb.connect('../capillary.db')

# Kör SQL-skript i ordning
for sql_file in ['create_database.sql', 'merge_llm.sql','read_comments.sql']: #
    print(f"Kör {sql_file}...")
    con.execute(open(f'{sql_file}').read())
    print(f"  -> klar")

df = con.execute(""" 
    SELECT *
    FROM protein_data 
    WHERE value IS NOT NULL
    AND analysis IS NOT NULL
    AND protein_value IS NOT NULL
    ORDER BY row_id
""").df()

df = pre_process_data(df)

con.execute("DROP TABLE IF EXISTS protein_data")
con.execute("CREATE TABLE protein_data AS SELECT * FROM df")

con.close()
print("Succesfully pre-processed the data")
```
